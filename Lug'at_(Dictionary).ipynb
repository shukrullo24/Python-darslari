#Lug'atlar bilan ishlash

Python izohli lug'atini yarating va lug'atga kamida 10 ta so'z qo'shing. Lug'atdagi har bir kalit va qiymatni for tsikli yordamida, alifbo ketma-ketligida chiroyli qilib konsolga chiqaring.

```python
cars = {
    "model":"ferrari",
    "rang":"qizil"
}

print("Birinchi ro'yxat:")
print(cars["rang"])
print(cars["model"])

en_uz = {
    "apple":"olma",
    "watermelon":"tarvuz",
    "melon":"qovun"
}

print("\nIkkinchi ro'yxat:")
print(en_uz["apple"])
print(en_uz["watermelon"])
print(en_uz["melon"])
```

```python
mevalar = {"olma":10000, "tarvuz":8000, "qovun":12000}
print(f"Olmaning narxi {mevalar["olma"]} so'm")
print(f"Tarvuz narxi {mevalar["tarvuz"]} so'm")
print(f"Qovun narxi {mevalar["qovun"]} so'm")
```
